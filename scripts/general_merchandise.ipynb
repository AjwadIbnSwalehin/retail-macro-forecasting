Import Libraries    

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

Output Table

```python
# Change the path to what suits your machine
df = pd.read_csv('c:/Users/Ajwad/OneDrive/Documents/repos/retail-macro-forecasting/data/general_merchandise_stores_adjusted.csv', index_col = 0)
print(df)
```

Create Plot

```python
yearly_averages = []

for year in df.index:
    average = df.loc[year].mean()
    yearly_averages.append(average)

plt.plot(df.index, yearly_averages)
plt.title("General Merchandise Stores Yearly Average 1992-2025")
plt.xlabel("Year")
plt.ylabel("Average Sales (Millions of Dollars)")
plt.show()
```
